# START – uruchomienie pakietu w Google Colab

Tego notebooka użyjemy tylko do uruchomienia materiałów w Colabie.

**Krok 1:** uruchom poniższą komórkę i wybierz plik
`eda_bank_marketing_colab.zip`.

```python
from google.colab import files
import io, zipfile
from pathlib import Path

uploaded = files.upload()
zip_name = next(name for name in uploaded if name.endswith(".zip"))

with zipfile.ZipFile(io.BytesIO(uploaded[zip_name])) as zf:
    zf.extractall("/content")

root = Path("/content/eda_bank_marketing_colab")
print("Pakiet rozpakowany:", root)
print("Następnie otwórz notebook: notebooks/00_EDA_wspolne.ipynb")
```
